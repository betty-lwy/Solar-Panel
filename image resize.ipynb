```python
import os
import cv2
import numpy as np
```

```python
def resize_and_pad(image, target_size):
    # Get the shape of the image
    h, w, c = image.shape
    scale = min(target_size / h, target_size / w)
    new_h, new_w = int(h * scale), int(w * scale)
    resized_image = cv2.resize(image, (new_w, new_h))
    
    # Create a new image with target size and pad the resized image onto it
    padded_image = np.zeros((target_size, target_size, c), dtype=np.uint8)
    # Center the resized image in the padded image
    pad_h = (target_size - new_h) // 2
    pad_w = (target_size - new_w) // 2
    padded_image[pad_h:pad_h + new_h, pad_w:pad_w + new_w, :] = resized_image
    return padded_image

def process_images(image_dir, output_dir, target_size):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    for image_file in os.listdir(image_dir):
        if image_file.endswith(('.jpg', '.jpeg', '.png')):
            image_path = os.path.join(image_dir, image_file)
            image = cv2.imread(image_path)
            if image is None:
                print(f"Failed to read image {image_file}")
                continue
            padded_image = resize_and_pad(image, target_size)
            output_path = os.path.join(output_dir, image_file)
            cv2.imwrite(output_path, padded_image)
            # print(f"Processed and saved {image_file}")
    return None
    
def label_generation(image_dir, annotation_dir):
    if not os.path.exists(annotation_dir):
        os.makedirs(annotation_dir)

    for image_file in os.listdir(image_dir):
        if image_file.endswith('.jpg'):
            # Corresponding annotation file
            annotation_file = os.path.splitext(image_file)[0] + '.txt'
            annotation_path = os.path.join(annotation_dir, annotation_file)
            
            # Check if the annotation file exists
            if not os.path.exists(annotation_path):
                # Create an empty annotation file
                open(annotation_path, 'w').close()
    return None
```

```python
# Image resize and padding
image_dir = 'path/to/your/images'
output_dir = 'path/to/your/annotations'
target_size = 640  

process_images(image_dir, output_dir, target_size)
```

```python
# Generate label files for images without objects
image_dir = 'path/to/your/images'
annotation_dir = 'path/to/your/annotations'

label_generation(image_dir, annotation_dir)
```
