```python
import geopandas as gpd
```

```python
footprints = gpd.read_file('original boundary file')
footprints = footprints[footprints['F_TYPE']== 'COMMON']
footprints['centroid'] = footprints.geometry.centroid.to_wkt()
footprints['OBJECTID'] = footprints['OBJECTID'].astype(int).astype(str)
```

```python
# Save the modified shapefile
footprints.to_file('edited boundary file')
```
