```python
from google.cloud import vision
import io
import os
```

```python
def detect_objects(image_path):
    try:
        with io.open(image_path, 'rb') as image_file:
            content = image_file.read()
    except Exception as e:
        print(f"Error reading image file {image_path}: {e}")
        return [], []

    image = vision.Image(content=content)

    try:
        # Perform object localization
        response = client.object_localization(image=image)
        objects = response.localized_object_annotations
    except Exception as e:
        print(f"Error calling Vision API: {e}")
        return [], []

    if not objects:
        print(f"No objects detected in {image_path}")
    else:
        print(f"Objects detected in {image_path}: {len(objects)}")

    # Print raw response for debugging
    print(response)

    residential_houses = [obj for obj in objects if 'house' in obj.name.lower() or 'residential building' in obj.name.lower()]
    solar_panels = [obj for obj in objects if 'solar panel' in obj.name.lower()]

    return residential_houses, solar_panels
```

```python
client = vision.ImageAnnotatorClient.from_service_account_json('your api json.json')

residential_houses, solar_panels = detect_objects('input_file')
for house in residential_houses:
    print(f"Residential house detected: {house.name} with confidence {house.score}")
for panel in solar_panels:
    print(f"Solar panel detected: {panel.name} with confidence {panel.score}")
```
